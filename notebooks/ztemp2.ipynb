```python
from halib import *

df = pd.read_csv(
    r"/mnt/e/SyncData/paper2_main/zcompare/all_new_no_temp/_cmp_raw_results.csv",
    sep=";",
    encoding="utf-8",
)
columns = df.columns.tolist()
stick_cols = columns[:3]  # first three columns
other_cols = [col for col in columns if "num_wrong_frames" in col]
cols = stick_cols + other_cols

df = df[cols]
context_df = pd.read_csv("./datasets/DFire_all_context.csv", sep=";", encoding="utf-8")

df = df.merge(context_df, on="video", how="inner")

# move last 3 columns to be first 3 columns
last_3_cols = df.columns[-3:].tolist()
df = df[last_3_cols + df.columns[:-3].tolist()]

df.to_csv(
    r"/mnt/e/SyncData/paper2_main/zcompare/all_new_no_temp/_cmp.csv",
    sep=";", encoding="utf-8", index=False)
```
