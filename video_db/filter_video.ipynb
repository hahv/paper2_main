```python
from halib.utils.video import VideoUtils
from halib.filetype import csvfile

df = VideoUtils.get_video_dir_meta_df(r"/mnt/c/Users/ha/Desktop/new_dataset/pool", video_exts=['.mp4', '.avi', '.mov', '.mkv', '.mpg'], search_recursive=True)
# csvfile.show(df)
# video_path	width	height	frame_count	fps
# /mnt/c/Users/ha/Desktop/new_dataset/pool/DFire/test/DFire__lb_firesmoke__VP1.mp4	1280	720	60	2
# Extract filename
df["file_name"] = df["video_path"].str.split("/").str[-1]

# Split by "__"
splits = df["file_name"].str.split("__", expand=True)

# Add dataset name and video label
df["dataset_name"] = splits[0]
df["video_label"] = splits[1].str.replace("lb_", "", regex=False)

df.to_csv(r"../zout/ds_pool.csv", index=False, sep=";")
```

```python
from halib import *
import os

os.chdir(r"/mnt/e/NextCloud/paper2_main")

df = pd.read_csv("./zout/ds_pool.csv", sep=";")

# HD = (width >= 1280) & (height >= 720)
filter1 = df["width"] >= 1280
filter2 = df["height"] >= 720
fps_filter = df["fps"] >= 10
frame_count_filter = df["frame_count"] >= 10


db_filter = df["dataset_name"] != "DFire"


df_filtered = df[filter1 & filter2 & fps_filter & frame_count_filter & db_filter]
df_filtered.to_csv(r"./zout/ds_pool_filtered.csv", index=False, sep=";")
```

```python
TARGET_DIR = r"/mnt/c/Users/ha/Desktop/new_dataset/pool_filtered"

for _, row in tqdm(df_filtered.iterrows()):
    src_path = row["video_path"]
    target_dir = os.path.join(TARGET_DIR, row["video_label"])
    os.makedirs(target_dir, exist_ok=True)

    dst_file_path = os.path.join(target_dir, row["file_name"])
    if not os.path.exists(dst_file_path):
        print(f"Copying {src_path} to {dst_file_path}")
        fs.copy_file(src_path, dst_file_path)
    else:
        print(f"File {dst_file_path} already exists. Skipping.")
```
