```python
from halib.research.perftb import *
from halib.research.perfcalc import PerfCalc

def exp_csv_filter_fn1(exp_file_name: str) -> bool:
    return ("__perf.csv" in exp_file_name) and ("per_frame" in exp_file_name)
def exp_csv_filter_fn2(exp_file_name: str) -> bool:
    return ("__perf.csv" in exp_file_name) and ("per_video" in exp_file_name)


map_mode_to_func = {
    "per_frame": exp_csv_filter_fn1,
    "per_video": exp_csv_filter_fn2,
}

for mode in map_mode_to_func.keys():
    pertb = PerfCalc.gen_perf_report_for_multip_exps(
        indir=r"/mnt/e/SyncData/paper2_main/zout/perf",
        exp_csv_filter_fn=map_mode_to_func[mode],
    )
    # pertb.plot(f"./zout/perf/compare_{mode}.svg", width=1920, height=600)
    pertb.to_csv(f"./zout/perf/compare_{mode}.csv")
```
