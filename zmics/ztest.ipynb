```python
from halib import *
BASE_DIR = "./zcompare/znew_model/non_static_temp_vs_notemp"
# SRC_DIR = "notemp_method_96.56"
# DST_DIR = "temp_method_96.56_static_only"

df = pd.read_csv("./datasets/DFire_all_context.csv", sep=";", encoding="utf-8")
test_df = df[df['set'] == 'test']

CAM_STATUS = "non_static"
target_df = test_df[test_df['status'] == CAM_STATUS]
pprint(len(target_df))

def mk_dir_based_on_cam(src, static_df):
    src_dir = os.path.join(BASE_DIR, src)
    dst_dir = os.path.join(BASE_DIR, f'{src}_{CAM_STATUS}_only')
    os.makedirs(dst_dir, exist_ok=True)
    # for non static video, we just copy from the source dir to dst dir
    # this case we do not process the video (with temp method) for non static video
    for idx, row in tqdm(static_df.iterrows()):
        pprint(f"Processing row {idx}: {row['video']}")
        video_name = row['video'] + '_out.mp4'
        csv_fname = row['video'] + '_results.csv'
        src_path = os.path.join(src_dir, video_name)
        assert os.path.exists(src_path), f"File not found: {src_path}"
        dst_path = os.path.join(dst_dir, video_name)
        fs.copy_file(src_path, dst_path)

        assert os.path.exists(src_path), f"File not found: {src_path}"
        src_path = os.path.join(src_dir, csv_fname)
        dst_path = os.path.join(dst_dir, csv_fname)
        fs.copy_file(src_path, dst_path)

mk_dir_based_on_cam("notemp_method_96.56", target_df)
mk_dir_based_on_cam("temp_method_96.56", target_df)
```
